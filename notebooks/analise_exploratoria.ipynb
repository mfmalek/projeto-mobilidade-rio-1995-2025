```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

df = pd.read_csv('../data/processed/passageiros_rio_tratado.csv.gz')

plt.figure(figsize=(12, 6))
plt.plot(df['ano'], df['onibus'], marker='o', label='Rodoviário (Ônibus)', color='#e11d48', linewidth=2)
plt.plot(df['ano'], df['metro'], marker='s', label='Ferroviário (Metrô)', color='#2563eb', linewidth=2)
plt.plot(df['ano'], df['trem'], marker='^', label='Ferroviário (Trem)', color='#059669', linewidth=2)
plt.plot(df['ano'], df['hidroviario'], marker='d', label='Hidroviário (Barcas)', color='#d97706', linewidth=2)
plt.plot(df['ano'], df['aeroviario'], marker='x', label='Aeroviário', color='#7c3aed', linewidth=2)

plt.title('Evolução do Movimento Médio Diário de Passageiros no Rio de Janeiro (1995-2025)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Ano', fontsize=12)
plt.ylabel('Passageiros Médios Diários (x1000)', fontsize=12)
plt.legend(title='Modais de Transporte', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

plt.savefig('../docs/assets/grafico_evolucao_modais.png', dpi=300, bbox_inches='tight')
plt.show()
```
